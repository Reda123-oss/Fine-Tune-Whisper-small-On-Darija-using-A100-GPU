```python
def normalize_ar(text):
    if not text:
        return ""
    text = text.replace("أ", "ا")
    text = text.replace("إ", "ا")
    text = text.replace("آ", "ا")
    text = text.replace("ّ", "")   # shadda
    text = text.replace("َ", "")   # fatha
    text = text.replace("ُ", "")   # damma
    text = text.replace("ِ", "")   # kasra
    text = text.replace("ْ", "")   # sukun
    text = text.replace("ً", "")   # tanwin fath
    text = text.replace("ٌ", "")   # tanwin damm
    text = text.replace("ٍ", "")   # tanwin kasr
    text = re.sub(r'[^\u0600-\u06FF\s]', '', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
import numpy as np
import librosa

def preprocess(example):
    try:
        audio = example["audio"]
        text = example["text"]

        # -------------------------
        # AUDIO LOAD
        # -------------------------
        audio_array = np.array(audio["array"], dtype=np.float32)
        sampling_rate = audio["sampling_rate"]

        if audio_array is None or len(audio_array) < 800:
            return None

        # -------------------------
        # RESAMPLE
        # -------------------------
        if sampling_rate != 16000:
            audio_array = librosa.resample(
                audio_array,
                orig_sr=sampling_rate,
                target_sr=16000
            )

        # -------------------------
        # NORMALIZE
        # -------------------------
        max_val = np.max(np.abs(audio_array))
        if max_val > 1e-6:
            audio_array = audio_array / max_val

        # -------------------------
        # TRIM (IMPORTANT for GPU stability)
        # -------------------------
        max_len = 16000 * 30  # 30 sec max
        audio_array = audio_array[:max_len]

        # pad short audio instead of dropping
        if len(audio_array) < 16000:
            audio_array = np.pad(audio_array, (0, 16000 - len(audio_array)))

        # -------------------------
        # FEATURES (Whisper)
        # -------------------------
        inputs = processor(
            audio_array,
            sampling_rate=16000
        )

        input_features = np.array(
            inputs.input_features[0],
            dtype=np.float32
        )

        # -------------------------
        # TEXT CLEANING
        # -------------------------
        text = normalize_ar(text)
        if text is None:
            text = ""

        text = text.strip()

        # DO NOT DROP TOO MUCH DATA
        if len(text) == 0:
            text = "silence"

        # -------------------------
        # TOKENIZATION
        # -------------------------
        labels = tokenizer(text).input_ids

        if len(labels) == 0:
            labels = [0]

        return {
            "input_features": input_features,
            "labels": labels
        }

    except Exception:
        # NEVER break dataset pipeline
        return {
            "input_features": np.zeros((80, 3000), dtype=np.float32),
            "labels": [0]
        }
```

```python
import torch, gc
from transformers import WhisperForConditionalGeneration
from peft import LoraConfig, get_peft_model, TaskType
import types

gc.collect()
torch.cuda.empty_cache()

# ── Load model ──
model = WhisperForConditionalGeneration.from_pretrained(
    "openai/whisper-medium",
    torch_dtype=torch.bfloat16,   # ✅ change float16 → bfloat16
    device_map={"": "cuda:0"},
    attn_implementation="eager",
)

model.config.use_cache = False
model.generation_config.forced_decoder_ids = None
model.generation_config.suppress_tokens = []
model.generation_config.language = "arabic"
model.generation_config.task = "transcribe"
model.enable_input_require_grads()

# ── Apply LoRA ──
lora_config = LoraConfig(
    r=64,
    lora_alpha=128,
    lora_dropout=0.1,
    bias="none",
    task_type=TaskType.SEQ_2_SEQ_LM,
    target_modules=["q_proj","k_proj","v_proj","out_proj","fc1","fc2"],
)
model = get_peft_model(model, lora_config)

# ════════════════════════════════════════
# PERMANENT FIX — patch the broken forward
# ════════════════════════════════════════
# ════════════════════════════════════════
# PERMANENT FIX — patch the broken forward
# ════════════════════════════════════════
import types

def _patched_forward(
    self,
    input_ids=None,
    attention_mask=None,
    inputs_embeds=None,
    decoder_input_ids=None,
    decoder_attention_mask=None,
    decoder_inputs_embeds=None,
    labels=None,
    output_attentions=None,
    output_hidden_states=None,
    return_dict=None,
    task_ids=None,
    **kwargs,
):
    # ── remove peft-specific args ──
    for key in list(kwargs.keys()):
        if hasattr(self, "special_peft_forward_args") and key in self.special_peft_forward_args:
            kwargs.pop(key)

    # ✅ FIX: remove input_features from kwargs if it exists
    # to avoid "multiple values" conflict
    input_features = kwargs.pop("input_features", None)

    # ✅ use input_features from kwargs if provided
    # otherwise use input_ids (what PEFT passes)
    actual_input = input_features if input_features is not None else input_ids

    return self.base_model(
        input_features=actual_input,
        attention_mask=attention_mask,
        decoder_input_ids=decoder_input_ids,
        decoder_attention_mask=decoder_attention_mask,
        decoder_inputs_embeds=decoder_inputs_embeds,
        labels=labels,
        output_attentions=output_attentions,
        output_hidden_states=output_hidden_states,
        return_dict=return_dict,
        **kwargs,
    )

# ── Apply patch ──
model.forward = types.MethodType(_patched_forward, model)
print("✅ Forward patched")

# ── Verify patch works ──
sample = dataset["train"][0]
test_features = torch.tensor(
    [sample["input_features"]], dtype=torch.bfloat16
).to("cuda:0")
test_labels = torch.tensor([sample["labels"]]).to("cuda:0")

with torch.no_grad():
    out = model(input_ids=test_features, labels=test_labels)
print(f"✅ Test forward pass OK — loss: {out.loss.item():.4f}")
print("✅ Model ready to train")
```

```python
from dataclasses import dataclass
from typing import Any, Dict, List, Union
import torch

@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any

    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        
        # ✅ 1. AUDIO FEATURES
        input_features = [
            {"input_features": f["input_features"]}
            for f in features
        ]
        batch = self.processor.feature_extractor.pad(
            input_features,
            return_tensors="pt"
        )

        # 🔥 ADD THIS LINE HERE
        batch["input_features"] = batch["input_features"].to(torch.bfloat16)

        # ✅ 2. LABELS
        label_features = [
            {"input_ids": f["labels"]}
            for f in features
        ]
        labels_batch = self.processor.tokenizer.pad(
            label_features,
            return_tensors="pt"
        )

        labels = labels_batch["input_ids"]

        # ✅ 3. MASKING PAD TOKENS
        labels = labels.masked_fill(
            labels_batch["attention_mask"] == 0,
            -100
        )

        batch["labels"] = labels

        return batch

# ✅ THIS LINE WAS MISSING — must instantiate it
data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor=processor)
print("✅ Data collator ready")
```

```python
import evaluate
import numpy as np

metric = evaluate.load("wer")

def compute_metrics(pred):
    pred_ids = pred.predictions
    label_ids = pred.label_ids

    # ✅ replace -100 BEFORE decoding (important)
    label_ids = np.where(label_ids != -100, label_ids, processor.tokenizer.pad_token_id)

    # ✅ decode
    pred_str = processor.tokenizer.batch_decode(
        pred_ids, skip_special_tokens=True
    )

    label_str = processor.tokenizer.batch_decode(
        label_ids, skip_special_tokens=True
    )

    # ✅ normalize (your function)
    pred_str = [normalize_ar(t) for t in pred_str]
    label_str = [normalize_ar(t) for t in label_str]

    # ✅ compute WER
    wer = metric.compute(
        predictions=pred_str,
        references=label_str
    )

    return {"wer": float(wer)}
```

```python
training_args = Seq2SeqTrainingArguments(
    output_dir=f"{BASE_DIR}/models/whisper-darija-a100",

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    gradient_accumulation_steps=4,

    num_train_epochs=3,
    learning_rate=3e-5,
    warmup_steps=500,
    weight_decay=0.01,

    # ✅ FIX — replace fp16 with bf16 for A100
    bf16=True,          # ← A100 supports bf16 natively, no grad scaling issues
    fp16=False,         # ← disable fp16

    optim="adamw_torch",
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={"use_reentrant": False},

    dataloader_num_workers=4,
    dataloader_pin_memory=True,

    evaluation_strategy="steps",
    eval_steps=500,
    save_steps=500,
    save_total_limit=3,
    logging_steps=25,

    predict_with_generate=True,
    generation_max_length=225,
    generation_num_beams=2,

    load_best_model_at_end=True,
    metric_for_best_model="wer",
    greater_is_better=False,

    remove_unused_columns=False,
    label_names=["labels"],
    save_safetensors=False,
    report_to="none",
)
```
